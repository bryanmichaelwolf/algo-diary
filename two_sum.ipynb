# Two Sum

## Description

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

### Example 1:
Input: nums = [2,7,11,15], target = 9
Output: [0,1]

### Example 2:
Input: nums = [3,2,4], target = 6
Output: [1,2]

### Example 3:
Input: nums = [3,3], target = 6
Output: [0,1]

## Solution

```python
/**
 * @param {number[]} nums
 * @param {number} target
 * @return {number[]}
 */
var twoSum = function(nums, target) {
    for (let i = 0; i < nums.length; i++) {
        for (let q = i + 1; q < nums.length; q++) {
            if (nums[i] + nums[q] === target) {
                return [i, q];
            }
        }
    }
};
```

## Thought Process

### What We Know

We know that `nums` will be an array of of integers (whole numbers with no trailing decimals), and we know that `target` will also be an integer.

We know that they don't want us to return the integers that sum to `target` themselves, but the indices of those integers.

We can not use the same element twice. So we can't sum an element with itself to achieve a correct sum.

We also allowed to assume that each test case will only have exactly one answer.

### Problem Processing

Based on the fact that we are given an array, and the problem is asking us to return indices of numbers that equal to the target number, my mind immediately gravitates toward looping as a possible solution method; because looping will allow me to traverse the array and access each integer stored in `nums`.

If I can access each integer in `nums` with a loop, I should be able to devise a way to compare integers (or in this case, the sum of two integers) to the value of `target`. This is the whole crux of the challenge, because if I find two integers in `nums` that sum to the value of `target` I need to return the indices of those integers.

### Execution

My solution is pretty basic and straightforward. I end up using two for loops. One nested inside the other.

My first for loop is relative standard. Its only purpose is to access each integer inside of `nums` in sequence; starting from the 0th index and moving forward through the array. I can now use the expression `nums[i]` to access an integer for comparison.

My nested for loop does all the heavy lifting. It is almost identical to the parent loop, except for the fact that I set the value of `q` to `i + 1`. This means that each time my "i loop" begins a new iteration, and it kicks off the "q loop", `nums[q]` will initiate as the array integer immediately after `nums[i]`.

The relationship of `i` and `q` is important for a couple reasons. Firstly, since `q` will always start out with a value of `i + 1`, this will prevent `nums[q]` from not only ever being the exact same array element as `nums[i]`, but also means that `nums[q]` will also never be an array element that `nums[i]` was in a previous iteration. This helps us eliminate the possibility of comparing the same element to itself and getting a false positive output. Secondly, because the "q loop" is nested inside of the "i loop", it completes multiple iterations for each single iteration of the "i loop". This means that the "q loop" can act as a "lookahead", in relation to our "i loop". So `nums[q]` can be used to look at every integer that comes after `nums[i]`. This set us up for a straightforward comparison of the values of `nums[i]` and `nums[q]` to determine if they equal `target`.

All that is left to do at this point is to execute conditional logic to say: if `nums[i] + nums[q]` equals `target`, return and array of the indices `i` and `q`.
